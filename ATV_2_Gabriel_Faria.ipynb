```python
from IPython.display import display
# Importando bibliotecas

#Analise de dados
import pandas as pd
import numpy as np

Valor_referencia = 85
#Tabela de dados
tabela = {
    "Sensor": [
        "Sensor S1",
        "Sensor S2",
        "Sensor S3",
        "Sensor S4",
        "Sensor S5",
        "Sensor S6",
    ],
    "Temp_Lida (°C)": [84.820, 85.150, 84.985, 85.040, 85.310, 84.910],
}

dados = pd.DataFrame(tabela)
display(dados)
```

Qualquer medida física obtida por um instrumento (como comprimento, temperatura, massa, tempo) é expressa por uma quantidade contínua que pode assumir valores inteiros, fracionários (racionais) ou com infinitas casas decimais não periódicas. O conjunto dos Números Reais engloba perfeitamente todas essas possibilidades de medidas físicas.

Os naturais e inteiros são muito restritos, pois ignoram os valores decimais e frações que qualquer medição real com boa precisão costuma apresentar

OS números complexos não fazem sentido físico para medidas de grandezas macroscópicas usuais, pois introduzem a unidade imaginária ($i$), que não tem correspondência direta na leitura de um sensor ou instrumento convencional
