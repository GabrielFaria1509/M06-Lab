## Contexto: Nós de uma Função de Onda Quântica

Na disciplina de F04 (Física Moderna e Quântica), os **nós de uma função de onda** representam posições nas quais a amplitude de probabilidade é nula. Em um modelo simplificado relacionado ao oscilador harmônico quântico, a parte polinomial de uma função de onda pode ser representada por:

$$\psi(x)=4x^3-6x$$

Para aplicar o método de Newton-Raphson, utiliza-se sua derivada:

$${\psi}'(x)=12x^2-6$$

Nesta atividade, o objetivo é localizar numericamente os valores de $x$ para os quais $\psi(x)=0$ e analisar como o chute inicial e a tolerância influenciam a convergência.

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 4*x**3 - 6*x

def df(x):
    return 12*x**2 - 6
```

```python
def newton_raphson(x0, tol=1e-9, max_iter=100):
    x = x0
    for i in range(max_iter):
        if abs(f(x)) < tol:
            return x, i
        if abs(df(x)) < 1e-12:
            raise ValueError("Derivada muito próxima de zero. Escolha outro chute inicial.")
        x = x - f(x)/df(x)
    return x, max_iter
```

1. O que significa afirmar que o método de Newton-Raphson possui convergência quadrática próximo de uma raiz simples?

Ao afirmarmos que o método de Newton-Raphson possui convergência quadrática próximo de uma raiz simples, significa que, perto de uma raiz simples, o núemro de dígitos corretos aproxidamente dobra a cada iteração. Desse modo, o erro de um a iteração para a próxima não diminui de forma linear, mas sim proporcional ao quadrado do erro anterior

2. Como o método pode ser adaptado quando a derivada da função não está disponível analiticamente?

Quando não se possui a expressão analítica da derivada da função, existem duas técnicas principais.A primeira técnica conssite em aproximar a derivada numericamente, ao invés de calcular diretamente a derivada, estima-se o valor deka usando dois pontos próximos, com uma diferença chamada de diferença finita progressiva.A segunda alternativa é pela técnica conhecida como método da secante, ele usa a inclinação da reta que passa por dois pontosanteriores da função como aproximação.

3. Compare o método de Newton-Raphson com o método da bissecção, destacando as vantagens e limitações de cada um.

Ambos os métodos servem para buscar as raízes, mas a bissecção se baseia em uma busca por intervalo, enquanto o Newton-Raphson é um método iterativo baseado na derivada.

O método da bisseção funciona dividindo repetidamente ao meio um intervalo [a,b] sabendo que f(a) e f(b) têm sianis opostos, ou seja, há uma raiz entre eles

Vantagens do método da bisseção

-Sempre converge, desde que o intervalo inicial esteja bem escolhido (com troca de sinal).

-Não precisa calcular a derivada.

-Mais simples de implementar e direto — risco muito menor de divergir.

Limtações do método da bisseção
-Convergência linear (lenta): o erro só é reduzido pela metade a cada iteração, então pode levar muitas iterações para atingir alta precisão

-Precisa de um intervalo inicial com troca de sinal — não funciona se a raiz for de multiplicidade par (a função toca o zero mas não muda de sinal)

Agora o método de Newton-Raphson funciona usando a reta tangente à função no ponto atual para estimar onde ela cruza o eixo x, repetindo esse processo a partir de um único chute inicial

Vantagens do método de Newton-Raphson

-Convergência quadrática perto da raiz (muito mais rápido que bisseção,uma vez que o número de dígitos corretos praticamente dobra a cada iteração)

-Não precisa de um intervalo com troca de sinal, só de um chute inicial.

Limitações do método de Newton-Raphson

-Precisa calcular ou aproximar a derivada a cada iteração

-Não garante convergência porque se o chute estiver longe da raiz ou a derivada for próxima de zero em algum ponto do caminho, o método pode divergir ou oscilar

-O método fica sensível à escolha dos chutes inciais, porque chutes diferentes podem levar raízes diferentes ou não convergir
